```python
import pandas as pd
import numpy as np

df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')
print(f"Jumlah baris awal dataset: {len(df)}")
df.head()
```

```python
jumlah_duplikat = df.duplicated(subset=['id_transaksi']).sum()
print(f"Jumlah baris duplikat berdasarkan id_transaksi: {jumlah_duplikat}")

df = df.drop_duplicates(subset=['id_transaksi'], keep='first').copy()
print(f"Jumlah baris setelah menghapus duplikat: {len(df)}")
```

```python
print("Jumlah Missing Value awal:")
print(df.isnull().sum())

df['kode_kasir'] = df['kode_kasir'].fillna('UNKNOWN')

print("\nJumlah Missing Value setelah ditanggulangi:")
print(df.isnull().sum())
```

```python
def clean_harga(val):
    if isinstance(val, str):
        val = val.replace('.', '').replace(',', '.')
    return float(val)

df['harga_satuan_clean'] = df['harga_satuan'].apply(clean_harga)
df['tanggal_clean'] = pd.to_datetime(df['tanggal'], format='mixed', dayfirst=True)
df['tahun'] = df['tanggal_clean'].dt.year
```

```python
df['total_pendapatan'] = (
    df['jumlah'] * df['harga_satuan_clean'] * (1 - (df['diskon_persen'] / 100.0))
)

df_selesai = df[df['status'].str.upper() == 'SELESAI']
pendapatan_2024 = df_selesai[df_selesai['tahun'] == 2024]['total_pendapatan'].sum()
pendapatan_2025 = df_selesai[df_selesai['tahun'] == 2025]['total_pendapatan'].sum()

print(f"Total Pendapatan Tahun 2024 : Rp {pendapatan_2024:,.2f}")
print(f"Total Pendapatan Tahun 2025 : Rp {pendapatan_2025:,.2f}")
```
